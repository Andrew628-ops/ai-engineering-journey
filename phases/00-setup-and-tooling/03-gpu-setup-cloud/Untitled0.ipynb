```python
import torch

print(torch.__version__)
print(torch.cuda.is_available())

torch.cuda.get_device_name(0)


a = torch.tensor([[1.0, 2.0]])
b = a.to("cuda")

print(a.device)
print(b.device)

size = 1000
a_cpu = torch.randn(size, size)
b_cpu = torch.randn(size, size)

a_gpu = a_cpu.to("cuda")
b_gpu = b_cpu.to("cuda")
print(a_cpu.device)
print(b_cpu.device)
print(a_gpu.device)
print(b_gpu.device)
```

```python
import time
result = a_cpu @ b_cpu
start = time.perf_counter()
for i in range(10):
  result = a_cpu @ b_cpu

elasped = time.perf_counter() - start
cpu_average = elasped / 10
print(cpu_average)
```

```python
import time
result = a_gpu @ b_gpu
torch.cuda.synchronize()
start = time.perf_counter()
for i in range(10):
  result = a_gpu @ b_gpu

torch.cuda.synchronize()
elasped = time.perf_counter() - start
gpu_average = elasped / 10
print(gpu_average)

speedup = cpu_average / gpu_average
print(speedup)
```

```python
properties = torch.cuda.get_device_properties(0)
print(properties.name)
result = properties.total_memory / 2
print(result / 1_000_000_000)
```
